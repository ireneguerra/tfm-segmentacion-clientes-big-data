```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from pathlib import Path
import numpy as np
import pandas as pd
from sklearn.preprocessing import PowerTransformer, StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def comparar_distribuciones(
    X_original,
    X_transformado,
    columns,
    transformation_name="transformación",
    bins=50
):

    for column in columns:

        if column not in X_original.columns:
            print(f"La columna '{column}' no existe en X_original.")
            continue

        if column not in X_transformado.columns:
            print(f"La columna '{column}' no existe en X_transformado.")
            continue

        fig, axes = plt.subplots(
            2,
            2,
            figsize=(14, 8)
        )

        # Histograma original
        sns.histplot(
            X_original[column],
            bins=bins,
            kde=True,
            ax=axes[0, 0]
        )

        axes[0, 0].set_title(
            f"{column} original"
        )
        axes[0, 0].set_xlabel(column)
        axes[0, 0].set_ylabel("Frecuencia")

        # Histograma transformado
        sns.histplot(
            X_transformado[column],
            bins=bins,
            kde=True,
            ax=axes[0, 1]
        )

        axes[0, 1].set_title(
            f"{column} después de {transformation_name}"
        )
        axes[0, 1].set_xlabel(column)
        axes[0, 1].set_ylabel("Frecuencia")

        # Boxplot original
        sns.boxplot(
            x=X_original[column],
            ax=axes[1, 0]
        )

        axes[1, 0].set_title(
            f"Boxplot de {column} original"
        )
        axes[1, 0].set_xlabel(column)

        # Boxplot transformado
        sns.boxplot(
            x=X_transformado[column],
            ax=axes[1, 1]
        )

        axes[1, 1].set_title(
            f"Boxplot de {column} después de "
            f"{transformation_name}"
        )
        axes[1, 1].set_xlabel(column)

        fig.suptitle(
            f"Comparación de la distribución de {column}",
            fontsize=14
        )

        plt.tight_layout()
        plt.show()
```

### Instacart

```python
instacart = pd.read_csv(
    "../../extended_datasets/instacart_model_extended.csv")
```

Se realizan transformaciones sobre aquellas variables que presentan una asimetría positiva elevada. Este tratamiento resulta especialmente relevante en métodos basados en distancias, ya que la presencia de colas largas y valores muy extremos puede provocar que determinados clientes ejerzan una influencia excesiva sobre la formación de los grupos. En las variables frequency, avg_basket_size, std_basket_size y unique_products se aplicó una transformación logarítmica mediante log1p(x), que permite comprimir los valores elevados manteniendo el orden relativo de las observaciones y, además, puede aplicarse cuando existen valores iguales a cero.

En el caso de unique_aisles, la transformación logarítmica resultó demasiado intensa y generaba una distribución excesivamente sesgada en sentido contrario. Por ello se utilizó una transformación mediante raíz cuadrada, que reduce la asimetría de forma más moderada y conserva mejor la estructura original de la variable. Las variables que no presentan una asimetría problemática se mantienen sin transformación.


```python
# Copia del dataset original
df_modelo = instacart.copy()

# Guardar el identificador fuera del modelo
user_ids = df_modelo["user_id"].copy()

X = df_modelo.drop(columns=["user_id"]).copy()

# Variables con asimetría positiva elevada o moderada
log_columns = [
    "frequency",
    "avg_basket_size",
    "std_basket_size",
    "unique_products"
]

sqrt_columns = [
    "unique_aisles"
]

# Aplicar log(1 + x) y sqrt(x) a las variables seleccionadas
X_transformado = X.copy()

X_transformado[log_columns] = np.log1p(
    X[log_columns]
)

X_transformado[sqrt_columns] = np.sqrt(
    X[sqrt_columns]
)


# Estandarizar todas las variables después de las transformaciones
scaler = StandardScaler()

X_scaled = pd.DataFrame(
    scaler.fit_transform(X_transformado),
    columns=X_transformado.columns,
    index=X_transformado.index
)


# Volver a añadir el identificador únicamente para guardar el resultado
instacart_scaled = pd.concat(
    [
        user_ids.reset_index(drop=True),
        X_scaled.reset_index(drop=True)
    ],
    axis=1
)

display(instacart_scaled.head())
```

```python
comparar_distribuciones(
    X_original=X,
    X_transformado=X_transformado,
    columns=log_columns,
    transformation_name="log1p"
)
```

```python
comparar_distribuciones(
    X_original=X,
    X_transformado=X_transformado,
    columns=sqrt_columns,
    transformation_name="sqrt"
)
```

```python
instacart_scaled.to_csv(
    "../../scaled_data/instacart_scaled.csv",
    index=False
)
```

### HyM

```python
ID_COLUMN = "customer_id"
hym_original = pd.read_csv(
    "../../extended_datasets/hym_model_extended.csv")
```

```python

hym_model = hym_original.copy()

# frequency ya representa el número de fechas de compra distintas
hym_model["single_purchase"] = (
    hym_model["frequency"] == 1
).astype("int8")
```

```python
df_modelo = hym_model.copy()
X = df_modelo.drop(columns=["customer_id"]).copy()
```

Se utilizó `log1p` en las variables con colas derechas pronunciadas, raíz cuadrada en `recency` al requerir una corrección más moderada y Yeo-Johnson en `monetary`, `avg_price` y `std_price` por su elevada asimetría.

Las variables que no requerían transformación se mantuvieron sin cambios y `single_purchase` se conservó como variable binaria. Finalmente, todas las variables continuas se estandarizaron mediante `StandardScaler` para situarlas en una escala comparable.

```python
log_columns = [
    "frequency",
    "avg_basket_size",
    "std_basket_size",
    "avg_days_between_purchases",
    "std_days_between_purchases",
    "unique_articles",
    "unique_product_types",
    "unique_sections",
    "unique_garment_groups"
]

sqrt_columns = [
    "recency"
]
# Estas variables tenían una fuerte asimetría,
# pero valores muy pequeños, por lo que usamos Yeo-Johnson
power_columns = [
    "monetary",
    "avg_price",
    "std_price"
]

# No se transforman, pero sí se escalan posteriormente
no_transform_columns = [
    "unique_product_groups",
    "article_variety_ratio",
    "channel_2_ratio"
]

# Variable binaria: se conserva como 0/1
binary_columns = [
    "single_purchase"
]

feature_columns = (
    log_columns
    + sqrt_columns       # Esto te faltaba
    + power_columns
    + no_transform_columns
    + binary_columns
)
```

```python
required_columns = [ID_COLUMN] + feature_columns

missing_columns = [
    column
    for column in required_columns
    if column not in hym_model.columns
]

if missing_columns:
    raise ValueError(
        f"Faltan estas columnas en el dataset: {missing_columns}"
    )


# Matriz de variables
X_original = hym_model[feature_columns].copy()

# Sustituir infinitos por NaN
X_original = X_original.replace(
    [np.inf, -np.inf],
    np.nan
)

# Comprobar nulos
null_counts = X_original.isna().sum()
null_counts = null_counts[null_counts > 0]

if not null_counts.empty:
    raise ValueError(
        "Se encontraron valores nulos:\n"
        f"{null_counts}"
    )

# log1p solo admite valores mayores o iguales que 0
negative_columns = [
    column
    for column in log_columns
    if (X_original[column] < 0).any()
]

if negative_columns:
    raise ValueError(
        "Estas columnas contienen valores negativos y no se puede "
        f"aplicar log1p directamente: {negative_columns}"
    )
```

```python
X_transformed = X_original.copy()


# Transformación logarítmica
# log1p(0) = 0, por lo que los clientes con una sola compra
# mantienen 0 en las variables de intervalos
X_transformed[log_columns] = np.log1p(
    X_original[log_columns]
)

# Raíz cuadrada
X_transformed[sqrt_columns] = np.sqrt(
    X_original[sqrt_columns]
)


# Transformación Yeo-Johnson para las variables de precio
price_transformer = PowerTransformer(
    method="yeo-johnson",
    standardize=False
)

X_transformed[power_columns] = (
    price_transformer.fit_transform(
        X_transformed[power_columns]
    )
)


# La columna binaria se mantiene como 0/1
X_transformed["single_purchase"] = (
    X_transformed["single_purchase"].astype("int8")
)
```

```python
scale_columns = [
    column
    for column in feature_columns
    if column not in binary_columns
]

scaler = StandardScaler()

X_scaled = X_transformed.copy()

X_scaled[scale_columns] = scaler.fit_transform(
    X_transformed[scale_columns]
)

# Mantener single_purchase sin escalar
X_scaled["single_purchase"] = (
    X_transformed["single_purchase"].astype("int8")
)
```

```python
comparar_distribuciones(
    X_original=X_original,
    X_transformado=X_transformed,
    columns=log_columns,
    transformation_name="log1p"
)
```

```python
comparar_distribuciones(
    X_original=X_original,
    X_transformado=X_transformed,
    columns=sqrt_columns,
    transformation_name="sqrt"
)
```

```python
comparar_distribuciones(
    X_original=X_original,
    X_transformado=X_transformed,
    columns=power_columns,
    transformation_name="Yeo-Johnson"
)
```

```python
hym_scaled = pd.concat(
    [
        hym_model[["customer_id"]].reset_index(drop=True),
        X_scaled.reset_index(drop=True)
    ],
    axis=1
)

display(hym_scaled.head())
```

```python
hym_scaled.to_csv(
    "../../scaled_data/hym_scaled.csv",
    index=False
)
```
