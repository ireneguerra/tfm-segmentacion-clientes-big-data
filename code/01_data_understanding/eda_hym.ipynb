### Descripción de los datos

El conjunto de datos **H&M Personalized Fashion Recommendations** contiene información anonimizada sobre clientes, artículos de moda y transacciones realizadas en H&M.

Los datos se distribuyen principalmente en las siguientes tablas:

- `articles.csv`
- `customers.csv`
- `transactions_train.csv`
- `sample_submission.csv`

Además, se proporciona una carpeta `images/` que contiene imágenes asociadas a una parte de los artículos. Pero para este proyecto no se abordarán las imágenes.

La estructura de los datos permite relacionar cada transacción con el cliente que realizó la compra y con el artículo adquirido.

---

#### 1. Tabla `articles`

La tabla `articles` contiene la información descriptiva de los productos o artículos comercializados por H&M.

Cada fila representa un **artículo diferente**.

##### Identificación del artículo

| Columna | Descripción |
|---|---|
| `article_id` | Identificador único del artículo. Es la clave que permite relacionar esta tabla con `transactions_train`. |
| `product_code` | Código del producto base al que pertenece el artículo. Distintos `article_id` pueden compartir un mismo `product_code` cuando representan variantes del mismo producto. |
| `prod_name` | Nombre del producto. |

Un `product_code` puede, por tanto, estar asociado a varios artículos concretos, por ejemplo diferentes variantes de color de un mismo producto.

---

##### Tipo de producto

| Columna | Descripción |
|---|---|
| `product_type_no` | Identificador numérico del tipo de producto. |
| `product_type_name` | Nombre del tipo de producto. Representa una categoría específica, como camiseta, pantalón, vestido, sujetador, etc. |
| `product_group_name` | Categoría más general a la que pertenece el producto, como prendas de la parte superior, prendas de la parte inferior, accesorios, zapatos o ropa interior. |

Estas variables proporcionan distintos niveles de agrupación del producto.

Por ejemplo:

`Garment Upper body → T-shirt`

donde `Garment Upper body` sería el grupo general y `T-shirt` un tipo más específico.

---

##### Apariencia gráfica

| Columna | Descripción |
|---|---|
| `graphical_appearance_no` | Código numérico que identifica la apariencia gráfica o patrón del artículo. |
| `graphical_appearance_name` | Nombre de la apariencia gráfica del artículo. Puede representar patrones como liso, rayas, estampado, cuadros, etc. |

Estas variables permiten estudiar las preferencias de los clientes respecto al diseño visual de las prendas.

---

##### Color

| Columna | Descripción |
|---|---|
| `colour_group_code` | Código numérico del grupo de color del artículo. |
| `colour_group_name` | Nombre del grupo de color, como negro, blanco, azul oscuro, beige, rojo, etc. |
| `perceived_colour_value_id` | Identificador numérico de la luminosidad o intensidad percibida del color. |
| `perceived_colour_value_name` | Nombre de la luminosidad percibida del color, por ejemplo `Dark`, `Light`, `Medium` o `Dusty Light`. |
| `perceived_colour_master_id` | Identificador numérico de una agrupación más general del color. |
| `perceived_colour_master_name` | Nombre de la familia general de color percibida, como negro, azul, blanco, beige, rojo, etc. |

El conjunto de datos contiene, por tanto, diferentes niveles de información sobre el color.

`colour_group_name` representa un color más específico, mientras que `perceived_colour_master_name` agrupa los colores en familias más generales.

---

##### Departamento

| Columna | Descripción |
|---|---|
| `department_no` | Identificador numérico del departamento al que pertenece el artículo. |
| `department_name` | Nombre del departamento. Representa una clasificación comercial relativamente específica utilizada por H&M. |

Los departamentos permiten diferenciar grupos de productos dentro de la estructura comercial de H&M.

---

##### Índices

| Columna | Descripción |
|---|---|
| `index_code` | Código del índice o categoría comercial del artículo. |
| `index_name` | Nombre del índice al que pertenece el artículo. |
| `index_group_no` | Identificador numérico del grupo de índices. |
| `index_group_name` | Nombre del grupo de índices. Representa una clasificación más general del público o línea comercial. |

Estas variables permiten identificar grandes líneas comerciales de H&M, como:

- Ladieswear
- Menswear
- Baby/Children
- Sport

La relación puede entenderse de forma jerárquica:

`index_group → index`

donde `index_group_name` es una categoría más general e `index_name` una subdivisión de esta.

---

##### Sección

| Columna | Descripción |
|---|---|
| `section_no` | Identificador numérico de la sección del producto. |
| `section_name` | Nombre de la sección a la que pertenece el artículo. |

La sección proporciona otro nivel de clasificación dentro del catálogo y permite diferenciar colecciones o grupos de productos más concretos.

---

##### Grupo de prendas

| Columna | Descripción |
|---|---|
| `garment_group_no` | Identificador numérico del grupo de prendas. |
| `garment_group_name` | Nombre del grupo de prendas, como vestidos, prendas de punto, pantalones, camisetas, accesorios, etc. |

Esta clasificación está especialmente relacionada con el tipo físico de prenda.

---

##### Descripción textual

| Columna | Descripción |
|---|---|
| `detail_desc` | Descripción textual detallada del artículo, incluyendo información sobre su diseño, materiales, forma o características. |

Esta variable puede contener valores nulos cuando no existe una descripción disponible para el artículo.

---

#### 2. Tabla `customers`

La tabla `customers` contiene información anonimizada sobre los clientes de H&M.

Cada fila representa un **cliente diferente**.

| Columna | Descripción |
|---|---|
| `customer_id` | Identificador único y anonimizado del cliente. Permite relacionar esta tabla con `transactions_train`. |
| `FN` | Indicador relacionado con la suscripción a Fashion News de H&M. Cuando toma el valor `1`, el cliente está asociado a esta opción. Existen numerosos valores nulos. |
| `Active` | Indicador binario relacionado con el estado activo del cliente. El valor `1` identifica los registros marcados como activos, mientras que otros registros aparecen como nulos. |
| `club_member_status` | Estado del cliente dentro del programa de miembros de H&M. Puede contener valores como `ACTIVE`, `PRE-CREATE` o `LEFT CLUB`. |
| `fashion_news_frequency` | Frecuencia con la que el cliente acepta recibir comunicaciones o noticias de moda de H&M. Puede contener categorías como `Regularly` o `NONE`. |
| `age` | Edad del cliente. Puede contener valores nulos. |
| `postal_code` | Código postal anonimizado del cliente. No representa directamente el código postal real, sino una versión anonimizada para proteger su privacidad. |

---

#### 3. Tabla `transactions_train`

La tabla `transactions_train` contiene el historial de compras realizado por los clientes.

Cada fila representa la compra de **un artículo por parte de un cliente en una fecha determinada**.

| Columna | Descripción |
|---|---|
| `t_dat` | Fecha en la que se realizó la transacción. |
| `customer_id` | Identificador del cliente que realizó la compra. Permite relacionar la transacción con `customers`. |
| `article_id` | Identificador del artículo comprado. Permite relacionar la transacción con `articles`. |
| `price` | Precio pagado por el artículo en esa transacción. |
| `sales_channel_id` | Identificador del canal utilizado para realizar la compra. El conjunto contiene dos canales de venta, codificados como `1` y `2`. |

Una misma combinación de cliente, artículo y fecha puede aparecer más de una vez. Esto no implica necesariamente que exista un duplicado incorrecto, ya que varias filas iguales pueden representar la compra de varias unidades del mismo artículo.

---

#### 4. Tabla `sample_submission`

La tabla `sample_submission` contiene el formato requerido para presentar las predicciones de la competición de Kaggle.

| Columna | Descripción |
|---|---|
| `customer_id` | Identificador del cliente para el que se deben generar recomendaciones. |
| `prediction` | Lista de `article_id` recomendados para ese cliente, ordenados según la predicción realizada por el modelo. |

El objetivo original de la competición era predecir qué artículos compraría cada cliente durante los **7 días posteriores al final del periodo de entrenamiento**.

Esta tabla no aporta nuevas características de los clientes, productos o transacciones y no es necesaria para realizar el análisis exploratorio o la segmentación de clientes, por lo que no será empleada para este proyecto.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")

sns.set_theme(style="whitegrid")

DATA_DIR = Path("../../data/hym/")
```

```python
articles = pd.read_csv(
    DATA_DIR / "articles.csv"
)

customers = pd.read_csv(
    DATA_DIR / "customers.csv"
)

transactions = pd.read_csv(
    DATA_DIR / "transactions_train.csv",
    parse_dates=["t_dat"]
)
```

```python
tablas = {
    "articles": articles,
    "customers": customers,
    "transactions": transactions
}

for nombre, df in tablas.items():
    print(
        f"{nombre}: "
        f"{df.shape[0]:,} filas y {df.shape[1]} columnas"
    )
```

```python
for nombre, df in tablas.items():

    print("\n" + "=" * 70)
    print(nombre.upper())
    print("=" * 70)

    display(df.head())
```

```python
for nombre, df in tablas.items():

    print("\n" + "=" * 70)
    print(nombre.upper())
    print("=" * 70)

    df.info()
```

```python
for nombre, df in tablas.items():

    nulos = pd.DataFrame({
        "nulos": df.isna().sum(),
        "porcentaje": (df.isna().mean() * 100).round(2)
    })

    nulos = nulos[nulos["nulos"] > 0].sort_values(
        "porcentaje",
        ascending=False
    )

    print("\n" + "=" * 70)
    print(f"VALORES NULOS - {nombre.upper()}")
    print("=" * 70)

    if nulos.empty:
        print("No existen valores nulos.")
    else:
        display(nulos)
```

```python
columnas_nulos_customers = [
    "Active",
    "FN",
    "fashion_news_frequency",
    "age",
    "club_member_status"
]

for columna in columnas_nulos_customers:
    print("\n" + "=" * 70)
    print(f"COLUMNA: {columna}")
    print("=" * 70)

    print(f"Nulos: {customers[columna].isna().sum():,}")
    print(f"Valores únicos: {customers[columna].nunique(dropna=False):,}")

    display(
        customers[columna]
        .value_counts(dropna=False)
        .to_frame("cantidad")
    )
```

#### Tratamiento de nulos

```python
# Solo falta un 0.39 %. Conservamos los artículos y marcamos
# que no existe descripción.
articles["detail_desc"] = (
    articles["detail_desc"]
    .fillna("UNKNOWN")
)


# FN y Active solo contienen 1 o NaN.
# Se transforman en variables binarias:
# 1 = sí
# 0 = no marcado / no consta

customers["FN"] = (
    customers["FN"]
    .fillna(0)
    .astype(int)
)

customers["Active"] = (
    customers["Active"]
    .fillna(0)
    .astype(int)
)


# fashion_news_frequency:
# los nulos se imputan con "NONE" (no recibe newsletter).

customers["fashion_news_frequency"] = (
    customers["fashion_news_frequency"]
    .fillna("NONE")
)


# club_member_status:
# Los valores desconocidos se mantienen como categoría propia.

customers["club_member_status"] = (
    customers["club_member_status"]
    .fillna("UNKNOWN")
)


# age:
# Guardamos primero si el valor original era nulo.

customers["age_missing"] = (
    customers["age"]
    .isna()
    .astype(int)
)

# Imputamos mediante la mediana
mediana_edad = customers["age"].median()

customers["age"] = (
    customers["age"]
    .fillna(mediana_edad)
)

print(f"Mediana utilizada para age: {mediana_edad}")
```

```python
print("Nulos restantes en articles:")
display(
    articles.isna().sum()[
        articles.isna().sum() > 0
    ]
)

print("\nNulos restantes en customers:")
display(
    customers.isna().sum()[
        customers.isna().sum() > 0
    ]
)
```

```python
columnas = [
    "Active",
    "FN",
    "fashion_news_frequency",
    "club_member_status",
    "age_missing"
]

for columna in columnas:
    print("\n" + "=" * 60)
    print(columna)
    print("=" * 60)

    display(
        customers[columna]
        .value_counts(dropna=False)
    )
```
